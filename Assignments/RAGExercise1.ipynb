### Implement Your Own Retriever

```python
import re


docs = [
    "RAG stands for Retrieval-Augmented Generation.",
    "It combines information retrieval and text generation.",
    "LangChain can be used to build RAG pipelines easily.",
    "Transformers use self-attention to handle sequential data."
]

query = "What is RAG?"

# TODO: Implement a function that returns the most relevant doc(s)
def retrieve_context(query, docs, k=1):
    # HINT: break query and docs into lowercase words
    # Count how many query words appear in each doc
    # Sort by score and return top-k docs
    query_words = set(query.lower().split())
    scores = []
    for doc in docs:
        doc_words = set(re.findall(r"\w+", doc.lower()))
        score = len(query_words & doc_words)  # Count common words
        scores.append((score,doc))
    scores.sort(key=lambda x: x[0],reverse=True)  # Sort by score in descending order
    return [doc for score, doc in scores[:k]]  # Return top-k docs


context = retrieve_context(query, docs)
print("Retrieved context:", context)
```

### Write the Prompt Template

```python
# TODO: Write a function that creates a final prompt
def make_prompt(context, question):
    """
    Returns a formatted text block for the LLM.
    Example:
    "Context: ... \nQuestion: ... \nAnswer:"
    """
    context = "\n".join(context)
    return f"Context: {context}\nQuestion: {question}\nAnswer:"

prompt_text = make_prompt(context, query)
print(prompt_text)
```

## Plug in a HuggingFace Model
Experiment with temperature (0.2, 0.8) and observe changes.

```python
from transformers import pipeline, set_seed

set_seed(42)

hf_gen = pipeline(
    task="text-generation",
    model="distilgpt2",
    max_new_tokens=50,
    do_sample=True,
    temperature=0.7,
    truncation=True,
    pad_token_id=50256,          # distilgpt2 eos token, silences a warning
    return_full_text=False,      # return only the generated continuation
)

response = hf_gen(prompt_text)[0]["generated_text"]
print(response.strip())
```

```python
for temp in (0.2, 0.8):
    out = hf_gen(prompt_text, temperature=temp)[0]["generated_text"]
    print(f"--- temperature = {temp} ---")
    print(out.strip(), "\n")

# Observation: low temperature (0.2) is more focused and repetitive;
# high temperature (0.8) is more diverse but drifts off-topic more often.
```

## Connect It All: Build a Mini RAG Function

```python
def mini_rag(query):
    # TODO: use your previous functions here
    context = retrieve_context(query, docs)
    prompt = make_prompt(context, query)
    result = hf_gen(prompt)[0]["generated_text"]
    return result

# Test your mini RAG
print(mini_rag("How do transformers work?"))
```
