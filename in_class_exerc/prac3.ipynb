Dependency Tree

```python
import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

text = "I will wear a white shirt on Tuesday."

doc = nlp(text)

# Print dependency information
for token in doc:
    print(
        f"{token.text:<15}"
        f"{token.dep_:<10}"
        f"{token.head.text}"
    )
```

```python
from spacy import displacy
#from IPython.display import HTML, display

# style="dep" tells displaCy to create a dependency parse visualization
# jupyter=False prevents spaCy from trying to automatically display
html = displacy.render(doc, style="dep", jupyter=False)


# Convert the HTML string into an HTML object and display it inside the Jupyter Notebook
display(HTML(html))
```

```python
# Import spaCy
import spacy

# Load English NLP model
nlp = spacy.load("en_core_web_sm")

# Example text
text = "Microsoft announced that it will invest 3 billion euros in Germany by 2026. The project will support AI research in Berlin and Munich. Satya Nadella said the company wants to expand cloud infrastructure in Europe."

# Process the text
doc = nlp(text)

# Print named entities
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
print(f"{'Entity':<20} {'Label':<10} {'Explanation'}")
print("-" * 50)

for ent in doc.ents:
    print(f"{ent.text:<20} {ent.label_:<10} {spacy.explain(ent.label_)}")
```

```python
from spacy import displacy
from IPython.display import HTML, display

# Create NER visualization as HTML
html = displacy.render(doc, style="ent", jupyter=False)

# Display it safely in Jupyter
display(HTML(html))
```

```python
# Import Hamming distance function
from scipy.spatial import distance

# Define two strings of equal length
string_1 = "test"
string_2 = "text"

# Calculate normalized Hamming distance
# Result = differing positions / total positions
hamming_distance = distance.hamming(
    list(string_1),
    list(string_2)
)

print("Normalized Hamming Distance:",hamming_distance)

# Convert normalized distance to actual number of differing characters
actual_distance = hamming_distance * len(string_1)

print("Actual Hamming Distance:", actual_distance)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "I love natural language processing.",
    "Natural language processing is fascinating.",
    "I enjoy learning about machine learning."
]   
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)  
print("Vocabulary:", vectorizer.vocabulary_)
print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Feature Names:", vectorizer.get_feature_names_out())
print("TF-IDF Matrix (dense representation):\n", tfidf_matrix.todense())    

tfidf_matrix = vectorizer.fit_transform(documents)
similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_score = similarity_matrix[0, 1]
print("Cosine Similarity Score:", similarity_score)
```
